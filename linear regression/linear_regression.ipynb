```python
import numpy as np

import matplotlib.pyplot as plt
```

```python
x=np.arange(1,11)
y=np.arange(1,11)
np.random.seed(2)
w=np.random.rand(1)[0]
b=np.random.rand(1)[0]

eta=0.01
epochs=1000

for i in range(epochs):
    yhat=(w*x)+b
    w=w-eta*(yhat-y)@x/len(x)
    b=b-eta*np.mean(yhat-y)
    
    if (0.5*np.mean((yhat-y)**2)):
        break
    
xvals=np.arange(1,11)

yvals=(w*xvals)+b

plt.plot(xvals,yvals,'ro')
```
